### 🔥 Practice (NumPy)

```python
# Create 2D array
arr = np.array([[10,20,30],
                [40,50,60]])
```

```python
# Matrix Multiplication
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

result = np.dot(a,b)
# or
result = a@ b
result
```

```python
# Column Mean
arr.mean(axis=0)
```

```python
# Normalize Data
arr=np.array([10,20,30,40])
norm = (arr - arr.mean())/arr.std()
norm
```

#### 🔥 Why NumPy is Faster Than List?
NumPy arrays are stored in contiguous memory.

Written in C internally.

Vectorized operations (no Python-level loops).

Less overhead per element.

Python list processes elements one by one.
NumPy processes many elements at once in optimized C code.”

## 🐼 HOUR 8–9: PANDAS

```python
# 1️⃣ Series
import pandas as pd

s = pd.Series([10,20,30])
s
```

```python
# 2️⃣ DataFrame
data = {
    'Name': ["a","b"],
    "Marks":[80,90]
}
df = pd.DataFrame(data)
df
```

```python
# 3️⃣ read_csv
df = pd.read_csv("data.csv")
df.head()
df.info()
df.describe()
```

```python
# 5️⃣ loc vs iloc
df.loc[0, "Name"]   # label-based
df.iloc[0, 1]       # index-based
```

```python
# 6️⃣ Filtering
df[df["Marks"] > 80]
```

```python
# 7️⃣ Groupby
df.groupby("Class")["Marks"].mean()
```

```python
# 8️⃣ Handling Missing Values
df.isnull()
df.dropna()
df.fillna(0)
```

```python
# 9️⃣ Apply
df["Marks"] = df["Marks"].apply(lambda x: x + 5)
```

```python
# 🔟 Sorting
df.sort_values("Marks", ascending=False)
```
