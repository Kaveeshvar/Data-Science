```python
# ✅ 3) Parameterized Queries (MOST IMPORTANT)
query = "SELECT * FROM students WHERE marks > %s"
cursor.execute(query, (80,))
rows = cursor.fetchall()
```

🔥 Why parameterized query is important?
Because it prevents SQL injection by separating SQL code from user input.

Explain simply:
“If someone types bad input like ' OR 1=1 --, parameterized queries treat it as data, not as SQL.”

Extra points:
It also helps with correctness (quotes/escaping) and can improve performance with query plans.

```python
# 🧪 Practice: Full CRUD Example
CREATE TABLE students (
  id INT AUTO_INCREMENT PRIMARY KEY,
  name VARCHAR(50),
  marks INT
);
```

```python
# 1️⃣ Insert Record
insert_query = "INSERT INTO students (name, marks) VALUES (%s, %s)"
cursor.execute(insert_query, ("Alice", 95))
conn.commit()
print("Inserted!")
```

```python
# 2️⃣ Fetch Records
# Fetch all:
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()
for row in rows:
    print(row)

# Fetch one:

cursor.execute("SELECT * FROM students WHERE id = %s", (1,))
row = cursor.fetchone()
print(row)
```

```python
# 3️⃣ Update Record
update_query = "UPDATE students SET marks = %s WHERE id = %s"
cursor.execute(update_query, (88, 1))
conn.commit()
print("Updated!")
```

```python
# 4️⃣ Delete Record
delete_query = "DELETE FROM students WHERE id = %s"
cursor.execute(delete_query, (1,))
conn.commit()
print("Deleted!")
```

```python
# ✅ Best Practice: Always Close
cursor.close()
conn.close()
```
