```python
# Variables & Methods: Instance vs. Class vs. Static
class Student:
    school_name = "Greenwood High"  # Class Variable

    def __init__(self, name):
        self.name = name  # Instance Variable

    # Instance Method
    def introduce(self):
        return f"Hi, I'm {self.name} from {self.school_name}."

    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name

    @staticmethod
    def add_numbers(a, b):
        return a + b
    
student1 = Student("Kavee")

print(Student.school_name) # Greenwood High
Student.change_school("Summit Academy")
print(student1.introduce()) # ...from Summit Academy
```

Instance Variable: A student's name (unique to them).

Class Variable: The school name (same for everyone).

Instance Method: A student "studying" (uses their specific name).

@classmethod: Changing the school name for everyone.

@staticmethod: A tool that doesn't need student data, like a calculator.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private variable (Encapsulation)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New Balance: {self.__balance}")

    def get_balance(self): # Public getter
        return f"Balance is hidden, but I can tell you it's safe!"
```

```python
acc1 = BankAccount("Kavee",600000)
acc1.deposit(50000)
```

Encapsulation: Keeping data safe. In a school, a student shouldn't be able to walk into the office and change their own GPA. We make it "private" using __.

Abstraction: Hiding complexity. You press a button to "Print Report Card." You don't need to know how the printer's gears work; you just need the result.

```python
# 🧬 Inheritance & Polymorphism
class Person:
    def work(self):
        return "Doing general work"

class Teacher(Person):
    def work(self): # Method Overriding
        return "Teaching Biology"

class Janitor(Person):
    def work(self): # Method Overriding
        return "Cleaning the hall"

# Polymorphism in action
staff_members = [Teacher(), Janitor()]
for member in staff_members:
    print(member.work())
```

Inheritance: A Principal is a type of Staff. They get all the Staff traits but add their own.

Polymorphism: "Many shapes." Both a Teacher and a Student have a daily_routine() method, but the teacher's routine is "Teach," and the student's is "Learn."

```python
# 💎 The Diamond Problem & MRO
class A:
    def label(self): print("Label A")

class B(A):
    def label(self): print("Label B")

class C(A):
    def label(self): print("Label C")

class D(B, C):
    pass

obj = D()
obj.label() # Outputs "Label B" because B is first in the list (D, B, C, A)
print(D.mro()) # Method Resolution Order (MRO) is the sequence in which Python searches for an attribute or method in a class hierarchy
```

In Python, if a class inherits from two parents who share a common ancestor, Python uses Method Resolution Order (MRO) to decide which method to call. It follows a "Left-to-Right, Depth-First" search.

```python
# 🐍 Dunder Methods (__str__, __repr__)
class Student:
    def __init__(self,name):
        self.name = name
    def __str__(self):
        return f"Student: {self.name}"
s=Student("Leo")
print(s)
```

These let you define how your object looks when printed.

__str__: User-friendly (for the parents).

__repr__: Developer-friendly (for the programmers).

```python
# 🛠 Practice: ATM Simulation
class ATM:
    def __init__(self,pin,balance=0):
        self.__pin=pin
        self.__balance = balance
    def check_balance(self,entered_pin):
        if entered_pin == self.__pin:
            return f"Balance: ${self.__balance}"
    def withdraw(self,amount,entered_pin):
        if entered_pin != self.__pin:
            return "Transaction Failed: Wrong PIN"
        if amount > self.__balance:
            return "Insufficient funds!"
        self.__balance -= amount
        return f"Withdraw ${amount}. Remaining: ${self.__balance}"
    def deposit(self,amount):
        if amount<500:
            return "Minimum depost amount is $500"
        self.__balance+=amount
        return f"Deposit successful. Balance: {self.__balance}"
my_atm = ATM(pin=1234,balance=600000)
print(my_atm.withdraw(50000,1234))
```
