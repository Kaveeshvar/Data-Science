```python
# 4. Factorial (iterative)
n=5
fact=1
for i in range(1,n+1):
    fact*=i
fact
```

```python
# 5. Factorial (recursive)
def facto(n):
    if n < 0:
        return "Factorial not defined for negative numbers"
    elif n == 0 or n == 1:
        return 1
    else:
        return n * facto(n - 1)
print(facto(5))
```

```python
# Fibonacci (loop)
n=5

if n==1:
    print(1)
elif n==2:
    print(1,1)
elif n>2:
    print(1,1,end=" ")
    for i in range(2,n):
        print(i,end=" ")
```

```python
# 7. Fibonacci (while)
n=15
n1,n2=0,1
fib=[]
count=0
while count < n:
    fib.append(n1)
    nth =n1+n2
    n1=n2
    n2=nth
    count+=1
print(fib)
```

```python
def fib(n):
    if n<=1:
        return n
    else:
        return fib(n-1) + fib(n-2)
for i in range(5):
    print(fib(i))
```

```python
# 8. Check prime
def isprime(n):
    for i in range(2,int(n**0.5) + 1):
        if n%i==0:
            return False
    return True
isprime(15)
```
