```python
# 1️⃣ Tk() and mainloop()
import tkinter as tk
```

```python
root = tk.Tk()
root.title("My App")
root.geometry("300x200")
# Creates the main window.
```

```python
root.mainloop()
# This keeps the window running and listening for events.
```

```python
# 2️⃣ Label
label = tk.Label(root,text = "Hello")
```

```python
label.config(text="New Text")
```

```python
label.pack()
```

```python
# 3️⃣ Entry
entry = tk.Entry(root)
entry.pack()

value = entry.get()
```

```python
# 4️⃣ Button  
root = tk.Tk()
root.title("My App")
root.geometry("300x200")

button = tk.Button(root, text='Click')
button.pack()
```

```python
root.mainloop()
```

```python
# 5️⃣ Pack vs Grid
# Pack -> Stacks widgets vertically or horizontally
# Grid -> Row-column layout
```

```python
# 6️⃣ StringVar
root = tk.Tk()
root.title("My App")
root.geometry("300x200")

name_var = tk.StringVar()

entry = tk.Entry(root, textvariable=name_var)
entry.pack()
root.mainloop()
print(name_var.get())
```

```python
# # 7️⃣ EVENT-DRIVEN PROGRAMMING (IMPORTANT EXPLANATION)
# In normal programs, code runs from top to bottom.
# In GUI programs, nothing happens until user clicks something.
# The program waits for events like button click or typing.
# When event happens, it triggers a function.
```

## 🔥 PRACTICE APPS

```python
# 1️⃣ Simple Addition App
import tkinter as tk

def add():
    num1 = int(entry1.get())
    num2 = int(entry2.get())
    result_label.config(text=f"Result: {num1 + num2}")

root = tk.Tk()
root.title("Addition App")

entry1 = tk.Entry(root)
entry1.pack()

entry2 = tk.Entry(root)
entry2.pack()

btn = tk.Button(root, text="Add", command=add)
btn.pack()

result_label = tk.Label(root, text="Result:")
result_label.pack()

root.mainloop()
```

```python
# 2️⃣ Login Form
import tkinter as tk

def login():
    username = username_entry.get()
    password = password_entry.get()
    
    if username == "admin" and password == "1234":
        status_label.config(text="Login Successful")
    else:
        status_label.config(text="Login Failed")

root = tk.Tk()
root.title("Login Form")

tk.Label(root, text="Username").pack()
username_entry = tk.Entry(root)
username_entry.pack()

tk.Label(root, text="Password").pack()
password_entry = tk.Entry(root, show="*")
password_entry.pack()

tk.Button(root, text="Login", command=login).pack()

status_label = tk.Label(root)
status_label.pack()

root.mainloop()
```

```python
# 3️⃣ Button Click Counter
import tkinter as tk

class CounterApp:
    def __init__(self, root):
        self.count = 0
        self.label = tk.Label(root, text="Count: 0")
        self.label.pack()
        
        self.button = tk.Button(root, text="Click", command=self.increase)
        self.button.pack()
    
    def increase(self):
        self.count += 1
        self.label.config(text=f"Count: {self.count}")

root = tk.Tk()
app = CounterApp(root)
root.mainloop()
```

### 🎯 Common Interview Questions

Q: What is mainloop?
→ It keeps application running and listens for events.

Q: What is command parameter?
→ It binds a function to a button click event.

Q: Difference between pack and grid?
→ pack is simple stacking; grid uses row-column layout.

Q: What is event-driven programming?
→ Code runs only when event occurs.

Q: Why use StringVar?
→ To connect GUI widgets and variables dynamically.
