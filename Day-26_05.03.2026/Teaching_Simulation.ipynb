4️⃣ What is OOPS?

OOPS is a way of organizing code using real-life objects.

Example:

- Class = Car blueprint

- Object = Actual car

```python
class Dog:
    def __init__(self, name):
        self.name = name
#  We describe dog once, then create many dogs.
```
