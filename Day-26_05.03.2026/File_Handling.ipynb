# 📂 File Handling

| Mode | Meaning                                    |
| ---- | ------------------------------------------ |
| "r"  | Read (error if file doesn't exist)         |
| "w"  | Write (creates new or overwrites existing) |
| "a"  | Append (creates if not exists)             |
| "rb" | Read binary                                |
| "wb" | Write binary                               |


```python
# 1️⃣ open() Function
file = open("data.txt", "a")
```

```python
file = open("data.txt", "r")
```

What happens if file doesn't exist?

In "r" → FileNotFoundError

In "w" → Creates file

In "a" → Creates file

```python
# 3️⃣ with Statement (VERY IMPORTANT)
with open('data.txt',"r") as file:
    content = file.read()
```

Why use with?

Because it automatically closes the file.

## 🔥 PRACTICE IMPLEMENTATIONS

```python
# 1️⃣ Read File and Count Words
def count_words(filename):
    with open(filename,"r") as f:
        text = f.read()
        words = text.split()
        return len(words)
count_words("output.txt")
```

```python
# 2️⃣ Write Prime Numbers to File
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i == 0:
            return False
        return True
with open('primes.txt','w') as f:
    for num in range(1,101):
        if is_prime(num):
            f.write(str(num)+"\n")
```

```python
# 3️⃣ Read CSV and Compute Average
import csv

def compute_average(filename):
    total=0
    count = 0

    with open(filename,"r") as f:
        reader = csv.DictReader(f)
        for row in reader:
            total += int(row['RANK'])
            count+=1
    return total/count
compute_average("admission.csv")
```

```python
# 4️⃣ Append Logs
from datetime import datetime

def log_message(message):
    with open('log.txt','a') as f:
        f.write(f'{datetime.now()} - {message}\n')
log_message("Haai srushti")
```

```python
# 5️⃣ JSON Save & Load Example
import json

data = {
    "students" :[
        {'name':"Kavee","marks":90},
        {'name':"srus","marks":91}
    ]
}

with open('students.json','w') as f:
    json.dump(data,f,indent=4)
with open('students.json','r') as f:
    loaded_data = json.load(f)
print(loaded_data)
```
